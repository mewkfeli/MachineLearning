### Загрузка обученной модели Teachable Machine

```python
from keras.models import load_model
from PIL import Image, ImageOps
import numpy as np
import os

# Настройка отображения numpy чисел
np.set_printoptions(suppress=True)

# Загрузка предварительно обученной модели
try:
    model = load_model("keras_model.h5", compile=False)
except Exception as e:
    print("Ошибка при загрузке модели")
    print(e)

# Загрузка файла с названиями классов
class_names = open("labels.txt", "r").readlines()
```

### Функция для предсказания класса изображения

```python
def predict_image(img_path):

    # Входные данных модели
    data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
    
    # Загрузка и предобработка изображения
    image = Image.open(img_path).convert('RGB')
    size = (224, 224)
    image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
    image_array = np.asarray(image)  # Конвертация в массив
    
    # Нормализация пикселей
    normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
    
    # Заполнение входного массива данными изображения
    data[0] = normalized_image_array
    
    prediction = model.predict(data)
    
    index = np.argmax(prediction)
    class_name = class_names[index].strip()  # Название класса 
    confidence_score = prediction[0][index]  # Оценка достоверности предсказания
    
    return class_name, confidence_score
```

### Проверка работы модели на изображениях из папок hot_dog_validation и not_hot_dog_validation

```python
validation_dirs = ['hot_dog_validation', 'not_hot_dog_validation']
results = []

# Обработка всех изображений в папках
for dir_name in validation_dirs:
    folder = os.path.join(dir_name)
    # Перебор всех файлов в папке
    for img_file in os.listdir(folder):
        img_path = os.path.join(folder, img_file) 
        # Получение предсказания для текущего изображения
        class_name, confidence = predict_image(img_path)
        # Сохранение результатов (файл, исходная папка, предсказанный класс, достоверность)
        results.append((img_file, dir_name, class_name, confidence))

for r in results:
    print(f'Файл: {r[0]}, Папка: {r[1]}, Предсказанный класс: {r[2]}, Достоверность: {r[3]:.2f}')
```
